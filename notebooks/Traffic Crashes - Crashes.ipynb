```python

import requests
import pandas as pd
import os

url = "https://data.cityofchicago.org/resource/85ca-t3if.csv"

params = {
    "$where": "crash_date >= '2025-01-01T00:00:00' AND crash_date < '2027-01-01T00:00:00'",
    "$limit": 5000000
}

output_file = "/content/Traffic_Crashes_Crashes_2025_2026.csv"

with requests.get(
    url,
    params=params,
    stream=True,
    timeout=180
) as response:
    response.raise_for_status()

    with open(output_file, "wb") as f:
        for chunk in response.iter_content(chunk_size=1024 * 1024):
            if chunk:
                f.write(chunk)

print("Download complete!")
print("File size (MB):",
      round(os.path.getsize(output_file) / 1024**2, 2))
```

```python

import pandas as pd

file_path = "/content/Traffic_Crashes_Crashes_2025_2026.csv"

df = pd.read_csv(file_path, low_memory=False)

print("Dataset shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
display(df.head())
```

```python

df["crash_date"] = pd.to_datetime(
    df["crash_date"],
    errors="coerce"
)

print("Records by year:")
print(df["crash_date"].dt.year.value_counts().sort_index())

print("\nEarliest crash:", df["crash_date"].min())
print("Latest crash:", df["crash_date"].max())

print("\nInvalid dates:", df["crash_date"].isna().sum())
```

```python

# Create a cleaning copy
df_clean = df.copy()

# Standardize column names
df_clean.columns = df_clean.columns.str.upper()

print("Column names standardized!")
print(df_clean.columns.tolist())

print("\nDataset shape:", df_clean.shape)
```

```python

print("Dataset shape:", df_clean.shape)

print("\nMissing values by column:")
print(
    df_clean.isnull()
    .sum()
    .sort_values(ascending=False)
    .to_string()
)

print("\nDuplicate rows:",
      df_clean.duplicated().sum())

print("\nDuplicate CRASH_RECORD_ID:",
      df_clean["CRASH_RECORD_ID"].duplicated().sum())

print("\nMissing CRASH_RECORD_ID:",
      df_clean["CRASH_RECORD_ID"].isna().sum())

print("\nData types:")
print(df_clean.dtypes.to_string())
```

```python

numeric_cols = [
    "POSTED_SPEED_LIMIT",
    "NUM_UNITS",
    "INJURIES_TOTAL",
    "INJURIES_FATAL",
    "CRASH_HOUR",
    "CRASH_MONTH",
    "CRASH_DAY_OF_WEEK",
    "LATITUDE",
    "LONGITUDE"
]

print("NUMERIC COLUMN SUMMARY:")
print(df_clean[numeric_cols].describe().to_string())

print("\nSUSPICIOUS VALUES:")

print("Speed limits below 0:",
      (df_clean["POSTED_SPEED_LIMIT"] < 0).sum())

print("Speed limits above 80:",
      (df_clean["POSTED_SPEED_LIMIT"] > 80).sum())

print("Speed limits equal to 0:",
      (df_clean["POSTED_SPEED_LIMIT"] == 0).sum())

print("Number of units below 1:",
      (df_clean["NUM_UNITS"] < 1).sum())

print("Negative injury counts:",
      (df_clean["INJURIES_TOTAL"] < 0).sum())

print("Crash hours outside 0-23:",
      (~df_clean["CRASH_HOUR"].between(0, 23)).sum())

print("Crash months outside 1-12:",
      (~df_clean["CRASH_MONTH"].between(1, 12)).sum())

print("Crash days outside 1-7:",
      (~df_clean["CRASH_DAY_OF_WEEK"].between(1, 7)).sum())

print("\nMissing coordinates:",
      df_clean["LATITUDE"].isna().sum())
```

```python

print("SPEED LIMIT DISTRIBUTION:")
print(
    df_clean["POSTED_SPEED_LIMIT"]
    .value_counts()
    .sort_index()
)

print("\nZERO COORDINATES:")

print("Latitude = 0:",
      (df_clean["LATITUDE"] == 0).sum())

print("Longitude = 0:",
      (df_clean["LONGITUDE"] == 0).sum())

print("Both coordinates = 0:",
      (
          (df_clean["LATITUDE"] == 0) &
          (df_clean["LONGITUDE"] == 0)
      ).sum())

print("\nSUSPICIOUS COORDINATES:")

# Broad geographic bounds around Chicago
invalid_coords = (
    df_clean["LATITUDE"].notna() &
    df_clean["LONGITUDE"].notna() &
    (
        ~df_clean["LATITUDE"].between(41.6, 42.1) |
        ~df_clean["LONGITUDE"].between(-88.0, -87.5)
    )
)

print("Coordinates outside Chicago-area bounds:",
      invalid_coords.sum())

display(
    df_clean.loc[
        invalid_coords,
        ["CRASH_RECORD_ID", "LATITUDE", "LONGITUDE"]
    ].head(10)
)
```

```python

import numpy as np

# Identify invalid coordinates
invalid_coords = (
    (df_clean["LATITUDE"] == 0) &
    (df_clean["LONGITUDE"] == 0)
)

print("Invalid coordinates found:", invalid_coords.sum())

# Replace invalid coordinates with missing values
df_clean.loc[
    invalid_coords,
    ["LATITUDE", "LONGITUDE", "LOCATION"]
] = np.nan

# Replace zero speed limits with missing values
zero_speed = df_clean["POSTED_SPEED_LIMIT"] == 0

print("Zero speed limits found:", zero_speed.sum())

df_clean.loc[
    zero_speed,
    "POSTED_SPEED_LIMIT"
] = np.nan

# Verify cleaning
print("\nRemaining zero coordinates:",
      (
          (df_clean["LATITUDE"] == 0) &
          (df_clean["LONGITUDE"] == 0)
      ).sum())

print("Remaining zero speed limits:",
      (df_clean["POSTED_SPEED_LIMIT"] == 0).sum())

print("\nMissing coordinates:",
      df_clean["LATITUDE"].isna().sum())

print("Missing speed limits:",
      df_clean["POSTED_SPEED_LIMIT"].isna().sum())

print("\nDataset shape:", df_clean.shape)
```

```python

columns_to_check = [
    "WEATHER_CONDITION",
    "LIGHTING_CONDITION",
    "ROADWAY_SURFACE_COND",
    "ROAD_DEFECT",
    "TRAFFIC_CONTROL_DEVICE",
    "DEVICE_CONDITION",
    "FIRST_CRASH_TYPE",
    "CRASH_TYPE",
    "PRIM_CONTRIBUTORY_CAUSE",
    "SEC_CONTRIBUTORY_CAUSE",
    "REPORT_TYPE",
    "DAMAGE"
]

for col in columns_to_check:
    print(f"\n===== {col} =====")
    print("Unique values:", df_clean[col].nunique())
    print(
        df_clean[col]
        .value_counts(dropna=False)
        .head(15)
    )
```

```python

# Select text columns, excluding identifiers and location
categorical_cols = df_clean.select_dtypes(
    include=["object", "string"]
).columns.difference([
    "CRASH_RECORD_ID",
    "LOCATION",
    "DATE_POLICE_NOTIFIED"
])

# Standardize text formatting
for col in categorical_cols:
    df_clean[col] = (
        df_clean[col]
        .astype("string")
        .str.strip()
        .str.upper()
        .replace("", pd.NA)
    )

print("Categorical formatting complete!")

print("\nDataset shape:", df_clean.shape)

print("\nMissing REPORT_TYPE:",
      df_clean["REPORT_TYPE"].isna().sum())

print("\nDuplicate CRASH_RECORD_ID:",
      df_clean["CRASH_RECORD_ID"].duplicated().sum())
```

```python

# Convert police notification date
df_clean["DATE_POLICE_NOTIFIED"] = pd.to_datetime(
    df_clean["DATE_POLICE_NOTIFIED"],
    errors="coerce"
)

print("POLICE NOTIFICATION DATES")

print("Missing/invalid notification dates:",
      df_clean["DATE_POLICE_NOTIFIED"].isna().sum())

print("Notifications before crash:",
      (
          df_clean["DATE_POLICE_NOTIFIED"] <
          df_clean["CRASH_DATE"]
      ).sum())

print("\nINJURY COUNT VALIDATION")

injury_cols = [
    "INJURIES_FATAL",
    "INJURIES_INCAPACITATING",
    "INJURIES_NON_INCAPACITATING",
    "INJURIES_REPORTED_NOT_EVIDENT"
]

# Only compare records with complete injury data
complete = df_clean[
    ["INJURIES_TOTAL"] + injury_cols
].notna().all(axis=1)

injury_sum = df_clean.loc[
    complete, injury_cols
].sum(axis=1)

mismatches = (
    injury_sum !=
    df_clean.loc[complete, "INJURIES_TOTAL"]
)

print("Records with complete injury data:",
      complete.sum())

print("Injury total mismatches:",
      mismatches.sum())

print("\nDataset shape:", df_clean.shape)
```

```python

from google.colab import files

print("FINAL VALIDATION")

print("Dataset shape:", df_clean.shape)

print("Duplicate CRASH_RECORD_ID:",
      df_clean["CRASH_RECORD_ID"].duplicated().sum())

print("Missing CRASH_RECORD_ID:",
      df_clean["CRASH_RECORD_ID"].isna().sum())

print("Missing CRASH_DATE:",
      df_clean["CRASH_DATE"].isna().sum())

print("Records outside 2025–2026:",
      (~df_clean["CRASH_DATE"].dt.year.isin([2025, 2026])).sum())

print("Remaining zero speed limits:",
      (df_clean["POSTED_SPEED_LIMIT"] == 0).sum())

print("Remaining zero coordinates:",
      ((df_clean["LATITUDE"] == 0) &
       (df_clean["LONGITUDE"] == 0)).sum())

# Export cleaned dataset
output_file = "Traffic_Crashes_Crashes_Cleaned_2025_2026.csv"

df_clean.to_csv(output_file, index=False)

print("\nCleaned dataset saved:", output_file)

files.download(output_file)
```
